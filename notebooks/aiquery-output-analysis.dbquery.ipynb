```sql
SELECT DISTINCT
    place_id,
    address,
    LTRIM(SPLIT_PART(address, ',', -3)) AS city, 
    cast(rohitb_demo.costa_mart.get_lat_long(city):latitude as double) AS latitude, 
    cast(rohitb_demo.costa_mart.get_lat_long(city):longitude as double) AS longitude,
    rating, 
    response:bag_drop_checkin:baggage_policy AS baggage_policy,
    response:bag_drop_checkin:wait_time AS wait_time,
    response:boarding_experience:delay_communication AS delay_communication,
    response:boarding_experience:gate_assignment AS gate_assignment,
    response:customer_service:fairness_of_policies AS fairness_of_policies,
    response:customer_service:staff_helpfulness AS staff_helpfulness,
    response:flight_operations:delay_reason AS delay_reason,
    response:flight_operations:waiting_on_paperwork AS waiting_on_paperwork,
    response:refund_change_policy:change_fee_fairness AS change_fee_fairness,
    response:refund_change_policy:refund_experience AS refund_experience,
    response:seat_comfort_services:seat_comfort AS seat_comfort,
    response:seat_comfort_services:snack_beverage_charges AS snack_beverage_charges,
    response:unexpected_charges:baggage_fees AS baggage_fees,
    response:unexpected_charges:booking_errors AS booking_errors,
    response:review_summary AS review_summary, 
    review
FROM 
    rohitb_demo.spirit_demo.spirit_google_reviews_predictions join rohitb_demo.spirit_demo.google_reviews using(review)
```
