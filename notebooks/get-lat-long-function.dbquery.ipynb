```sql
CREATE OR REPLACE FUNCTION rohitb_demo.costa_mart.get_lat_long(location STRING)
  RETURN
    (
      WITH location_cte AS (
        SELECT
          parse_json(
            ai_query(
              'databricks-meta-llama-3-3-70b-instruct',
              'What is the latitude and longitude of ' || location || ', estimates are fine.',
              responseFormat =>
                '{
                "type": "json_schema",
                "json_schema": {
                  "name": "Location",
                  "schema": {
                    "type": "object",
                    "properties": {
                      "latitude": {"type": "number"},
                      "longitude": {"type": "number"}
                    }
                  }
                }
              }'
            )
          ) AS location
      )
      SELECT
        location
      FROM
        location_cte
    );

select rohitb_demo.costa_mart.get_lat_long('New York'):latitude
```
